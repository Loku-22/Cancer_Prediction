## DATA PROCESSING

```python
df = df.drop(columns=['Patient_ID'])
```

```python
df.shape
```

```python
x = df.drop(columns = ['Survival_Prediction'])
y = df['Survival_Prediction']
```

```python
x.shape
```

```python
from sklearn.preprocessing import LabelEncoder
categorical_cols = x.select_dtypes(include=['object']).columns
label_encoders = {} 
for col in categorical_cols:
    le = LabelEncoder()
    x[col] = le.fit_transform(x[col])
    label_encoders[col] = le 
```

```python
categorical_cols,len(categorical_cols)
```

```python
x.info()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
## CHI-SQUARE-TEST
from sklearn.feature_selection import chi2, SelectKBest
x_cat = x_train.select_dtypes(include=['int64','float64'])
chi2_selector = SelectKBest(score_func=chi2,k="all")
chi2_selector.fit(x_cat,y_train)
```

```python
chi2_scores = pd.DataFrame({
    'Feature': x_cat.columns,
    'Chi2_Score': chi2_selector.scores_
}).sort_values(by='Chi2_Score', ascending=False)
```

```python
chi2_scores
```

```python
top_features = (chi2_scores.head(5))["Feature"].tolist()
```

```python
top_features
```

```python
x = x[top_features]
```

```python
x.shape
```

```python
y.shape
```

## Model Training

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
x_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train,scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
gb = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1,max_depth=3,random_state=42)
```

```python
gb.fit(x_train,y_train)
```

```python
y_pred = gb.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,roc_auc_score,classification_report
accuracy = accuracy_score(y_test,y_pred)
```

```python
accuracy
```

```python
report = classification_report(y_test,y_pred)
```

```python
print(report)
```

```python
## ROC_AUC works only for binary classification
y_proba = gb.predict_proba(x_test)[:,1] if len(y.unique()) == 2 else None
roc_auc_score(y_test,y_proba)
```
